## 3. 데이터셋: PlantVillage

- 식물 잎 이미지를 보고 병해 종류를 분류하는 과제입니다
- https://github.com/spmohanty/plantvillage-dataset
- 원본은 **38개 클래스, 약 54,000장**입니다
- **단, 이 프로젝트는 클래스당 150장(총 약 5,700장)으로 줄인 고정 부분집합을 사용합니다.**
  - 이유 1: 전체를 쓰면 단순 CNN으로도 정확도가 95%+가 나와 **기법 간 차이가 안 보입니다**.
    데이터를 줄이면 기법 효과가 그래프에 뚜렷이 드러납니다.
  - 이유 2: 학습 시간이 짧아져 2주 안에 여러 실험을 반복할 수 있습니다.
  - 구글 다운로드 링크: https://drive.google.com/file/d/1oOe-OAYWOneBxADfABdZ8MVc7kq6zriO/view?usp=drive_link
- **`PER_CLASS=150`과 `SEED`는 절대 바꾸지 마세요.** 모든 팀이 동일한 데이터로
  실험해야 결과를 서로 비교할 수 있습니다 (담당 기법으로 바꾸는 부분만 변경).
- Starter 노트북에 다운로드 + 축소 코드가 포함되어 있습니다 (Kaggle 계정 불필요)

실제 기업 활용 사례: 농가에서 스마트폰으로 잎을 찍으면 병해를 자동 진단하는 앱

### 4-3. 상호 검증 및 Q&A — 댓글 (6/4(목) 23:59)

**검증 대상 배정 (고정 순환)**: 같은 분반 내에서 **N팀 → N+1팀**을 검증합니다.
번호가 가장 큰 팀은 **1팀**을 검증합니다. (예: 3팀 → 4팀, 마지막 9팀 → 1팀)

- **[팀 필수]** 배정된 팀(N+1)의 코드를 받아 **처음부터 순서대로 실행**,
  오류 없이 재현되는지 결과를 댓글로 보고 (팀원 1인이 대표로)
- **[개인 필수]** 본인이 **서로 다른 3개 팀**의 코드를 실행하고 각 팀에
  질문(오류·결과 해석 등)을 댓글로
- **[개인 선택]** 4개 팀 이상에 추가 질문 (점수 가산 아님, 학습·교류 권장)

> **질문은 실명으로 답니다.** 단, **예의 바른 비판적 질문은 공격이 아니라 평가
> 대상**입니다. "이 결과가 왜 이렇게 나왔는지 모르겠다"류의 날카로운 질문이
> 형식적 칭찬보다 높은 점수를 받습니다. 받는 팀도 감정적으로 받아들이지 마세요
> — 답변의 충실도(C-2)로 점수를 회복할 기회입니다.
> *(LMS가 익명 댓글을 지원하고 교수자가 원할 경우, 동료에게는 익명·교수자에게만
> 실명으로 운영할 수 있습니다. 기본은 실명입니다.)*

### 4-4. 질문에 대한 답변 — 댓글 (6/7(일) 23:59)

- **[팀 필수]** 우리 팀에 달린 **모든 질문에** 답글 (팀원들이 나눠서 답해도 되며,
  누가 어떤 질문에 답했는지 드러나야 C-2 개인 평가가 됩니다)
- **[개인 선택 · 재질문 허용]** 받은 답변이 불충분하다고 판단되면 **재질문 1회**를
  달 수 있습니다. 재질문에 대한 재답변도 6/7까지 가능하면 답해 주세요.
  (전면 의무는 아니며, 부실 답변을 짚는 재질문은 받는 팀의 C-2 평가에 영향)

## 5. 평가 기준 (루브릭) — 총 100점, 절대평가

> **accuracy가 낮아도 실험을 성실하게 했으면 A를 받을 수 있습니다.**
> 이 프로젝트는 "좋은 숫자"가 아니라 **"비교·해석·고찰의 깊이"** 를 평가합니다.

---

### A. 실험 과정 (40점) — 이 프로젝트의 핵심

**A-1. Baseline 재현 (8점)**

| 점수 | 기준 |
|------|------|
| 8 | Starter 베이스라인이 오류 없이 실행되고 결과가 노트북에 기록됨 |
| 4 | 실행은 됐으나 결과 기록 누락 |
| 0 | 베이스라인 재현 불가 |

**A-2. 기법 비교 실험 (16점)**

| 점수 | 기준 |
|------|------|
| 16 | 최소 3개 기법(비-TL 1개 이상) 실험 + 각각 "왜 했는가·결과 수치"가 기록됨 |
| 11 | 3개 이상이나 일부 이유/수치 누락, 또는 비-TL 조건 미충족 |
| 6 | 2개만 시도 |
| 2 | 1개만 시도 |
| 0 | 베이스라인 외 추가 실험 없음 |

**A-3. 비교·해석·고찰 (16점)** ← 가장 비중 있는 항목

| 점수 | 기준 |
|------|------|
| 16 | 기법 간 결과를 **한 그래프/표로 비교**하고, "왜 이 기법이 더(덜) 좋았는지"를 **데이터 근거를 들어** 자신의 언어로 해석함. 성능이 안 오른 실험도 원인을 분석함 |
| 11 | 비교는 했으나 해석이 일반론에 그침("augmentation은 보통 좋다" 식) |
| 6 | 수치 나열만 있고 비교/해석이 약함 |
| 0 | 결과만 있고 해석 없음 |

> **복붙 방지**: 코드는 검색·LLM 보조가 허용되지만, **"왜 이런 결과가 나왔는지"
> 해석은 베낄 수 없습니다.** 출처 있는 기법을 그대로 적용했더라도, 그 결과를
> 우리 데이터·우리 그래프로 설명하지 못하면 A-3에서 점수를 받지 못합니다.
> 반대로 목록 안의 평범한 기법만 썼어도 해석이 깊으면 만점입니다.

### C. Q&A 참여 (20점) — 모두 **개인 평가**

**C-1. 질문 참여 (10점)**

| 점수 | 기준 |
|------|------|
| 10 | 서로 다른 **3개 팀**에 질문 댓글을 달았음 |
| 6 | 2개 팀에 질문 |
| 3 | 1개 팀에 질문 |
| 0 | 미참여 |

> 질문받은 팀이 최종 제출에 **가장 도움이 된 질문**으로 꼽으면
> (4-5 참고) 강사 재량으로 일부 가산될 수 있습니다.

> **좋은 질문 예 (권장, 점수 강제 아님)** — "ResNet에서 val acc가 epoch 10에
> 급등했는데, lr scheduler step과 우연히 겹친 건지 구조 덕분인지 구분되나요?
> scheduler를 꺼서 다시 돌려보면 확인되지 않을까요?"
> "Augmentation을 넣었는데 test가 거의 그대로인데, train-val 격차는 줄었나요?
> 줄었다면 '효과 없음'이 아니라 '과적합은 줄었으나 정확도 이득이 작다'로
> 해석해야 하지 않나요?"
>
> **형식적 질문 예** — "26번째 줄에 오타 있어요" / "그래프 축 라벨이 없네요" /
> "수고하셨어요". (이런 질문만 있어도 C-1 점수는 동일하지만, 받은 팀의
> 추천을 받지 못하고 본인 학습에도 도움이 안 됩니다.)

**C-2. 답변의 충실도 (10점)** — 답변한 개인별 평가 (복수 답변 가능)

| 점수 | 기준 |
|------|------|
| 9~10 | 질문의 핵심을 정확히 짚고, 데이터·코드 근거나 추가 확인 결과로 답함. 모르면 "확인 못 했다"고 솔직히 + 어떻게 알아볼지 제시 |
| 6~8 | 자신의 언어로 답했으나 근거가 약하거나 일부만 답함 |
| 3~5 | 답변 시도했으나 핵심에서 벗어남 / "그냥 그렇게 했습니다"류 |
| 0 | 답변 없음 |

> 받은 답이 부실하면 질문자가 **재질문 1회**를 달 수 있습니다(4-4). 재질문에
> 성실히 재답변하면 C-2를 회복할 수 있고, 무응답이면 낮은 구간으로 평가됩니다.

## 6. 실험 주제 — 메뉴와 선택 규칙

### 선택 규칙

- **최소 3개 기법**을 골라 baseline과 비교합니다.
- 그중 **최소 1개는 비-TL 계열**이어야 합니다 (TL만 3개는 불가).
- baseline(SimpleCNN, 기본 Normalize)은 **공통 출발점**으로 전 팀 동일하게 실행합니다.
- 같은 노력이라도 화려한 기법을 고르는 것보다 **비교·해석을 깊게** 하는 것이
  점수에 더 중요합니다 (루브릭 A-3).

### 기법 메뉴

**비-TL 계열 (이 중 최소 1개 필수)**

| 무엇을 바꾸나 | 참고 주차 |
|--------------|----------|
| augmentation 조합·강도 | 10주차 |
| CNN 채널 수·depth·block 구성 | 5주차 |
| optimizer·lr·scheduler | 7주차 |
| Residual 구조 도입 | 10주차 |
| Dropout·weight decay·BN | 6주차 |
| LR 스케줄 단독 비교 | 7주차 |

**TL 계열**

| 코드 | 무엇을 바꾸나 | 참고 주차 | 심화 예시(선택, 가점 없음) |
|------|--------------|----------|---------------------------|
| TL-FE | 사전학습 ResNet, backbone freeze | 10주차 | freeze 깊이별 비교 |
| TL-FT | 사전학습 ResNet, 단계적 unfreeze + differential LR | 10주차 | 더 깊은 unfreeze, discriminative LR |

> **심화 예시는 권장하지만 필수가 아니며, "새 기법을 썼다"는 사실 자체로는
> 가점이 없습니다.** 목록 안 기법만 써도 해석이 깊으면 만점입니다. 반대로
> 심화 기법을 가져와도 우리 데이터로 해석하지 못하면 점수가 없습니다.
> (취지: 신기법 베껴오기 경쟁 방지, 해석 능력 우선)

### 공정 비교 원칙

- **한 번에 한 가지만** 바꾸고 `PER_CLASS`·`SEED`·`EPOCHS`·batch size 등
  나머지는 고정합니다.
- 모든 기법 결과를 **하나의 비교 그래프/표**로 모아 제시합니다.
- 각 실험마다 "왜 이 결과가 나왔는가"를 반드시 한 단락 이상 서술합니다.

### 참고: 기법별 예상 경향 (강사 참고 실험 기준, 데이터·시드에 따라 다를 수 있음)

| 기법 | 경향 | 한 줄 해석 포인트 |
|------|------|-------------------|
| Baseline | ~81% | 출발점. 가벼운 과적합이 보임 |
| AUG | baseline과 비슷 | 과적합은 줄지만 이 데이터에선 정확도 이득은 작음 — "왜?"가 핵심 |
| RES | ~91% | 구조의 힘. 큰 도약 지점 |
| TL-FE | ~84% | freeze 한계 — from-scratch보다 낮을 수 있음 |
| TL-FT | ~93% | 적응시켜야 사전학습의 진짜 힘 |

> 위 표는 "정답"이 아니라 해석 토론용입니다. 본인 팀 결과가 다르면
> **그 차이의 원인을 분석**하는 것이 더 높은 점수로 이어집니다.

## 7. 자주 묻는 질문

**Q. 인터넷 검색이나 ChatGPT 사용해도 되나요?**
코드 작성 보조는 허용합니다. 단, **"왜 이 코드를 썼는지, 왜 이런 결과가 나왔는지"
설명할 수 없으면** 게시판 Q&A·루브릭 A-3에서 점수를 받지 못합니다. 코드는 베낄 수
있어도 해석은 베낄 수 없습니다.

**Q. 어떤 실험을 해야 하나요? 배정이 없나요?**
배정은 없습니다. [6번](#6-실험-주제-메뉴와-선택-규칙) 메뉴에서 최소 3개를
직접 고르되, 그중 1개 이상은 Transfer Learning 이외 방식이어야 합니다.

**Q. 목록에 없는 최신 기법을 써도 되나요?**
환영하지만 필수가 아니고, 쓴다는 사실만으로 가점은 없습니다. 평범한 기법이라도
비교·해석이 깊으면 만점이고, 화려한 기법이라도 해석이 없으면 점수가 없습니다.

**Q. 어떤 팀의 코드를 검증·질문하나요?**
같은 분반에서 **우리 팀 번호 +1 팀**을 팀 차원에서 검증하고(마지막 팀은 1팀),
개인은 서로 다른 3개 팀에 질문합니다. (4-3 참고)

**Q. 비판적인 질문을 하면 상대 팀이 기분 나빠하지 않을까요?**
질문 점수(C-1)는 "3개 팀에 질문했는가"로만 매겨지므로 날카롭게 묻든 가볍게
묻든 점수는 같습니다. 다만 좋은 질문은 받은 팀이 최종 제출에서 "가장 도움된
질문"으로 꼽아 소액 가산될 수 있고, 무엇보다 본인 학습에 도움이 됩니다.
받는 팀도 답변의 충실도(C-2)로 점수를 얻을 기회이니 감정적으로 받지 마세요.

**Q. 답변이 너무 부실한 팀이 있으면요?**
재질문 1회를 달 수 있습니다(4-4). 재질문에 무응답이면 그 팀의 C-2가 낮게
평가됩니다. (질문 점수 C-1은 참여 여부로만 매겨지므로 재질문이 본인 점수를
올리지는 않지만, 학습과 상대 팀 평가에 의미가 있습니다.)

**Q. 팀원이 참여를 안 해요.**
노트북의 개인 기여 섹션, 그리고 게시판 Q&A는 **개인 단위로** 평가되므로
참여하지 않은 팀원은 C 점수를 받지 못합니다.

**Q. 베이스라인보다 성능이 낮게 나와요.**
실험 기록과 해석이 충실하면 감점 없습니다. "왜 낮게 나왔는가"를 분석한 내용이
오히려 A-3에서 좋은 점수를 받습니다.

**Q. 다른 팀과 코드를 공유해도 되나요?**
게시판 공개 코드를 참고·검증하는 것이 이 과제의 핵심입니다. 단, 남의 코드를
자기 것으로 제출하는 복사는 금지입니다.

**Q. PER_CLASS나 SEED를 바꿔도 되나요?**
안 됩니다. 모든 팀이 동일 데이터로 비교해야 하므로 고정입니다. 담당 기법으로
바꾸는 부분만 변경하세요.
